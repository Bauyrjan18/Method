# CICIoT2023 Sample Data Exploration and Feature Analysis

This notebook provides quick validation of the micro-sample extracted from the CICIoT2023 benchmark dataset.
It inspects class distributions, extracts top-10 flow features, and verifies data integrity for edge intrusion detection.

```python
import pandas as pd
import yaml

# Load experiment configuration
with open('../configs/config.yaml', 'r') as f:
    config = yaml.safe_load(f)

print("Config loaded successfully:", config['experiment']['name'])
```

```python
# Load sample data
sample_path = '../' + config['data']['sample_path']
df = pd.read_csv(sample_path)
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Class distribution
print("Class distribution:")
print(df['label'].value_counts())

# Binary label mapping
binary_targets = (df['label'] != 'Benign').astype(int)
print("Binary targets (0: Benign, 1: Attack):\n", binary_targets.value_counts())
```

```python
# Top-10 features inspection
top_10 = config['feature_selection']['top_10_features']
print("Top-10 Features Summary:")
df[top_10].describe().T[['mean', 'std', 'min', 'max']]
```
